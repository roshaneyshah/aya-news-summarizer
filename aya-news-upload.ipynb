```python
import glob, os
from kaggle_secrets import UserSecretsClient
from huggingface_hub import HfApi, whoami

TOKEN = UserSecretsClient().get_secret("HF_Token_2")
api = HfApi(token=TOKEN)
user = whoami(token=TOKEN)["name"]
data = os.path.dirname(glob.glob("/kaggle/input/**/data/results.json", recursive=True)[0])
root = os.path.dirname(data)
print("Found:", root, os.listdir(root))

jobs = [(data, f"{user}/aya-news-curation", "dataset"),
        (f"{root}/adapter-raw", f"{user}/aya-expanse-8b-news-raw", "model"),
        (f"{root}/adapter-curated", f"{user}/aya-expanse-8b-news-curated", "model")]
for folder, repo, kind in jobs:
    api.create_repo(repo, repo_type=kind, exist_ok=True)
    api.upload_folder(folder_path=folder, repo_id=repo, repo_type=kind)
    print("UPLOADED: https://huggingface.co/" + ("datasets/" if kind == "dataset" else "") + repo)
print("DONE")
```
